```python
import pandas as pd
from pathlib import Path
import soundfile as sf
```

```python
data_path = Path('../data').resolve()
wav_files = list(data_path.rglob('*.wav'))

print(f"Hay {len(wav_files)} WAV files")
```

```python
wav_files[0]
```

```python

sr_l = []
ch_l = []
bd_l = []
for wav_file in wav_files:
    # y, sr = librosa.load(wav_file, sr=None, mono=False, duration=0.001)
    info = sf.info(wav_file)
    sr = info.samplerate
    channels = info.channels
    bitdepth = info.subtype_info
    sr_l.append(sr)
    ch_l.append(channels)
    bd_l.append(bitdepth)

    # for attr, value in info.__dict__.items():
        # print(f"{attr}: {value}")

    if sr!=48_000 or channels!=1 or bitdepth!='32 bit float':
        print(f"\nFile: {wav_file.name}")
        print(f"Sampling Rate: {sr} Hz")
        print(f"channels {channels}")
        print(f"bitdepth: {bitdepth}")
```

```python
df = pd.DataFrame(data={'wav_files': wav_files, 'sr': sr_l, 'channels': ch_l, 'bitdepth': bd_l})
```
